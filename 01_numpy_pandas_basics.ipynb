## NumPy — Quickstart

- Array là cấu trúc dữ liệu chính, nhanh và hiệu quả về bộ nhớ so với list thuần.
- Chú ý `shape`, `dtype`, và phép biến đổi (`reshape`, `astype`).

Thử các ví dụ dưới đây để nắm bắt cách tạo và kiểm tra properties của array.


```python
# Tạo và kiểm tra arrays
a = np.array([1, 2, 3])
b = np.arange(0, 10, 2)
c = np.linspace(0, 1, 5).reshape(5, 1)

print("a:", a, "shape:", a.shape, "dtype:", a.dtype)
print("b:", b, "shape:", b.shape)
print("c:\n", c)

# ép kiểu
float_a = a.astype(float)
print("float_a dtype:", float_a.dtype)
```

## Indexing, Slicing & Boolean Masks

- Slicing giống với list: `[start:stop:step]`.
- Boolean mask cho phép chọn phần tử theo điều kiện.
- Fancy indexing (bằng array chỉ số) trả về một copy.

Lưu ý quan trọng: slicing có thể là _view_ — thay đổi nó có thể ảnh hưởng tới array gốc.


```python
x = np.arange(12)
print("x =", x)
print("x[2:8:2] =", x[2:8:2])
print("x[-3:] =", x[-3:])
mask = x % 3 == 0
print("mask:", mask)
print("x[mask] =", x[mask])

# Fancy indexing
idx = np.array([0, 4, 7])
print("x[idx] =", x[idx])

# Example: views vs copies
s = x[2:6]
print("slice s before", s)
s[0] = 99
print("after modifying s -> x becomes", x)
# để tránh thay đổi gốc: use .copy()
cop = x[2:6].copy()
```

## Broadcasting & Vectorized Operations

- Broadcasting: phép toán giữa arrays có shape khác nhưng tương thích theo luật broadcasting.
- Vectorized operations: viết phép toán trên toàn bộ array thay vì loop — ngắn, nhanh.


```python
A = np.array([[1, 2, 3], [4, 5, 6]])
b = np.array([1, 0, 1])
print("A:\n", A)
print("b:", b)
print("A + b -> broadcasting:\n", A + b)

# Ví dụ vectorized: chuẩn hoá (z-score)
arr = np.random.randn(1000)
zscore = (arr - arr.mean()) / arr.std()
print("zscore mean ~", zscore.mean(), "std ~", zscore.std())
```

## Vectorized vs Loop — So sánh hiệu năng

Sử dụng `%timeit` để so sánh thao tác vectorized (NumPy) so với list comprehension/for-loop.


```python
N = 2_000_000
x = np.random.rand(N)
y = np.random.rand(N)

# Vectorized
print("Vectorized add:")
%timeit x + y

# Python loop
print("\nPython loop (list comprehension):")
%timeit [x[i] + y[i] for i in range(N)]
```

## Pandas — Series & DataFrame cơ bản

- `Series` là 1D, có label index.
- `DataFrame` là 2D, bảng dữ liệu với cột có tên.
- Học cách tạo, xem `head()`, chọn cột/ô, và `loc`/`iloc`.


```python
# Series
s = pd.Series([10, 20, 30], index=["a", "b", "c"])
print("Series:\n", s)

# DataFrame
df = pd.DataFrame({"name": ["Alice", "Bob", "Cara"], "age": [25, 30, 22], "score": [85, 90, 88]})
print("\nDataFrame:\n", df)

# Chọn hàng/cột
print("\ndf.loc[0] ->", df.loc[0].to_dict())
print('df["age"] ->', df["age"].values)

# Thêm cột mới vectorized
df["age_plus_1"] = df["age"] + 1
print("\nAfter add column:\n", df)
```

## Một số trick hay sử dụng thường xuyên 🔧

- Dùng `np.where` hoặc `np.select` để tạo cột mới theo điều kiện (vectorized).
- Dùng `Series.astype('category')` cho cột có ít giá trị để tiết kiệm bộ nhớ.
- Tránh dùng `.apply` khi có thể viết bằng vectorized ops.
- Kiểm tra dtype bằng `df.dtypes` và convert bằng `pd.to_numeric` / `pd.to_datetime` khi cần.


```python
# Một vài minh hoạ
arr = np.array([1, 2, 3, 4, 5])
print("np.where:", np.where(arr % 2 == 0, "even", "odd"))

df2 = pd.DataFrame({"color": ["red", "blue", "red", "green"]})
df2["color_cat"] = df2["color"].astype("category")
print("\ndf2 dtypes:\n", df2.dtypes)

# Vectorized conditional column
df2["is_red"] = np.where(df2["color"] == "red", True, False)
print("\ndf2:\n", df2)
```

## Bài tập ngắn (Exercises) 🧩

1. (NumPy) Viết hàm `top_k(arr, k=3)` trả về k phần tử lớn nhất từ một mảng 1D (không sort toàn bộ mảng).
2. (NumPy) Cho 2 arrays có shape khác nhau nhưng broadcast được — hãy mô tả shape sau khi broadcasting.
3. (Pandas) Tạo DataFrame mẫu, tính z-score cho cột `score` mà không dùng vòng lặp.
4. (Pandas) Dùng `np.where` để tạo cột boolean từ điều kiện trên câu 3.

Hãy thử tự làm trước khi xem đáp án. Các lời giải mẫu ở cell tiếp theo.


```python
# Solutions (mẫu)
import numpy as np
import pandas as pd


# Bài 1
def top_k(arr, k=3):
    arr = np.asarray(arr)
    if k <= 0:
        return np.array([])
    idx = np.argpartition(arr, -k)[-k:]
    return arr[idx][np.argsort(-arr[idx])]


print("top_k example:", top_k(np.array([5, 1, 3, 2, 9, 7]), k=3))

# Bài 2 (ví dụ shapes)
a = np.ones((3, 1))
b = np.array([1, 2, 3])
print("\nBroadcasted shapes: a", a.shape, "b", b.shape, "-> result", (a + b).shape)

# Bài 3 & 4 (Pandas)
df_ex = pd.DataFrame({"score": [50, 80, 90, 60, 70]})
df_ex["zscore"] = (df_ex["score"] - df_ex["score"].mean()) / df_ex["score"].std()
df_ex["high"] = np.where(df_ex["zscore"] > 0, True, False)
print("\nExample df_ex:\n", df_ex)
```
